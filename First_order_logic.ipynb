```python
def is_variable(x):
    return isinstance(x, str) and x.islower()

def apply_substitution(x, subst):
    if isinstance(x, str) and x in subst:
        return apply_substitution(subst[x], subst)
    return x

def unify(x, y, subst):
    x = apply_substitution(x, subst)
    y = apply_substitution(y, subst)

    if x == y:
        return subst

    if is_variable(x):
        if occurs(x, y, subst):
            return None
        subst[x] = y
        return subst

    if is_variable(y):
        if occurs(y, x, subst):
            return None
        subst[y] = x
        return subst

    if isinstance(x, tuple) and isinstance(y, tuple):
        if x[0] != y[0] or len(x) != len(y):
            return None

        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)
            if subst is None:
                return None
        return subst

    return None

def occurs(var, term, subst):
    term = apply_substitution(term, subst)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs(var, t, subst) for t in term[1:])

    return False


# Main program
# Test Case 1
p1 = ("Eats", "x", "Apple")
p2 = ("Eats", "Riya", "y")
result = unify(p1, p2, {})
print("Test Case 1:", result)

# Test Case 2
p1 = ("Likes", "x", "Mango")
p2 = ("Likes", "John", "Mango")
result = unify(p1, p2, {})
print("Test Case 2:", result)

# Test Case 3
p1 = ("P", "x", "a")
p2 = ("Q", "b", "y")
result = unify(p1, p2, {})
print("Test Case 3:", result)

# Test Case 4
p1 = ("Knows", "x")
p2 = ("Knows", ("f", "x"))
result = unify(p1, p2, {})
print("Test Case 4:", result)

```
