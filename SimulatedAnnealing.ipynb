```python
import random
import math

N = 8

def conflicts(board):
    count = 0
    for i in range(N):
        for j in range(i + 1, N):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                count += 1
    return count

def display(board):
    for row in range(N):
        print(" ".join("Q" if board[col] == row else "." for col in range(N)))

def run_simulated_annealing():
    board = [random.randint(0, 7) for _ in range(N)]
    T = 100.0
    alpha = 0.95
    step = 1

    while T > 0.01:
        current = conflicts(board)
        if current == 0:
            return board

        new_board = board.copy()
        col = random.randint(0, 7)
        new_board[col] = random.randint(0, 7)
        new_conflicts = conflicts(new_board)
        delta = new_conflicts - current

        print(f"\nStep {step} - Neighbor Board:")
        display(new_board)

        if delta <= 0:
            board = new_board
        elif random.random() < math.exp(-delta / T):
            board = new_board

        T *= alpha
        step += 1
    return board if conflicts(board) == 0 else None

def simulated_annealing_with_restarts():
    attempt = 1
    while True:
        print(f"--- Attempt {attempt} ---")
        solution = run_simulated_annealing()
        if solution is not None:
            print("\nSolution Found!")
            display(solution)
            break
        else:
            print("\nSolution not found, resetting...\n")
        attempt += 1

simulated_annealing_with_restarts()
```
