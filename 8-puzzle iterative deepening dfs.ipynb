```python
class EightPuzzleSolver:
    def __init__(self, start_state, goal_state=(1, 2, 3, 4, 5, 6, 7, 8, 0)):
        self.start_state = tuple(start_state)
        self.goal_state = tuple(goal_state)
        self.moves = [
            (-1, 0, 'Up'),
            (1, 0, 'Down'),
            (0, -1, 'Left'),
            (0, 1, 'Right')
        ]

    def get_neighbors(self, state):
        neighbors = []
        blank_idx = state.index(0)
        row, col = blank_idx // 3, blank_idx % 3

        for dr, dc, move_name in self.moves:
            new_row, new_col = row + dr, col + dc
            if 0 <= new_row < 3 and 0 <= new_col < 3:
                new_idx = new_row * 3 + new_col
                state_list = list(state)
                state_list[blank_idx], state_list[new_idx] = state_list[new_idx], state_list[blank_idx]
                neighbors.append((tuple(state_list), move_name))
        return neighbors

    def depth_limited_dfs(self, current_state, depth_limit, path):
        if current_state == self.goal_state:
            return path

        if depth_limit <= 0:
            return None

        for neighbor, move in self.get_neighbors(current_state):
            if neighbor not in [p[0] for p in path]:
                path.append((neighbor, move))
                result = self.depth_limited_dfs(neighbor, depth_limit - 1, path)
                if result is not None:
                    return result
                path.pop()

        return None

    def solve(self, max_depth=50):
        for depth in range(max_depth):
            initial_path = [(self.start_state, "Start")]
            result = self.depth_limited_dfs(self.start_state, depth, initial_path)
            if result is not None:
                return result
        return None

    def print_solution(self, path):
        if not path:
            return
        for step, (state, move) in enumerate(path):
            print(f"Step {step}: {move}")
            for i in range(0, 9, 3):
                print(state[i:i+3])
            print("-" * 15)

if __name__ == "__main__":
    initial_board = (
        1, 2, 3,
        4, 0, 6,
        7, 5, 8
    )
    
    solver = EightPuzzleSolver(start_state=initial_board)
    solution_path = solver.solve()
    solver.print_solution(solution_path)
```
