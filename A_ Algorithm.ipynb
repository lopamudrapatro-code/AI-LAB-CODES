```python
import heapq

# Goal state
goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)

# Heuristic 1: Misplaced Tiles
def misplaced_tiles(state):
    count = 0
    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1
    return count


# Heuristic 2: Manhattan Distance
def manhattan_distance(state):
    distance = 0
    for i in range(9):
        if state[i] != 0:
            goal_index = goal.index(state[i])

            current_row = i // 3
            current_col = i % 3

            goal_row = goal_index // 3
            goal_col = goal_index % 3

            distance += abs(current_row - goal_row)
            distance += abs(current_col - goal_col)

    return distance


# Generate neighboring states
def get_neighbors(state):
    neighbors = []
    zero = state.index(0)

    row = zero // 3
    col = zero % 3

    directions = [(-1, 0), (1, 0),
                  (0, -1), (0, 1)]

    for dr, dc in directions:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_index = new_row * 3 + new_col

            new_state = list(state)

            new_state[zero], new_state[new_index] = (
                new_state[new_index], new_state[zero]
            )

            neighbors.append(tuple(new_state))

    return neighbors


# A* Search Algorithm
def astar(initial, choice):

    if choice == 1:
        heuristic = misplaced_tiles
    elif choice == 2:
        heuristic = manhattan_distance
    else:
        print("Invalid choice")
        return

    open_list = []
    heapq.heappush(open_list,
                   (heuristic(initial), 0, initial))

    g_cost = {initial: 0}
    parent = {initial: None}

    while open_list:

        f, g, current = heapq.heappop(open_list)

        if g > g_cost[current]:
            continue

        if current == goal:
            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()

            print("\nSolution Path:")
            for state in path:
                print_state(state)

            print("Total moves:", len(path) - 1)
            return

        for neighbor in get_neighbors(current):
            new_g = g + 1

            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g
                parent[neighbor] = current

                h = heuristic(neighbor)
                f = new_g + h

                heapq.heappush(open_list,
                               (f, new_g, neighbor))

    print("No solution found")


# Display puzzle
def print_state(state):
    for i in range(0, 9, 3):
        print(*state[i:i+3])
    print()


# Main program
initial = (2, 8, 3,
           1, 6, 4,
           7, 0, 5)

print("A* Search - 8 Puzzle")
print("1. Misplaced Tiles")
print("2. Manhattan Distance")

choice = int(input("Enter your choice (1 or 2): "))

astar(initial, choice)
```
